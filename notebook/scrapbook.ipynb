```python
import pandas as pd
import glob
import os

# Load all data files from ../data/ directory
data_dir = '../data/'
dataframes = {}

for file_path in glob.glob(os.path.join(data_dir, '*')):
    file_name = os.path.basename(file_path)
    
    # Determine file type and load accordingly
    if file_path.endswith('.csv'):
        dataframes[file_name] = pd.read_csv(file_path)
    elif file_path.endswith('.parquet'):
        dataframes[file_name] = pd.read_parquet(file_path)
    elif file_path.endswith('.json'):
        dataframes[file_name] = pd.read_json(file_path)
    elif file_path.endswith('.xlsx'):
        dataframes[file_name] = pd.read_excel(file_path)

# Display loaded dataframes
for name, df in dataframes.items():
    print(f"{name}: {df.shape}")
```
